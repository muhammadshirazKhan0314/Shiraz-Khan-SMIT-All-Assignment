## 🧱 47. Largest Square Tile

### Goal:
    A floor is 48 ft by 18 ft. Find the biggest square tile that fits without cutting.

### Steps:

    Ask for the length and the width.
    
    Use a while loop with Euclid's method: while b is not 0, set a, b = b, a % b.
    
    The answer is the final value of a (the GCD).
    
    Print the tile size.

### Example:

    Length: 48
    Width: 18
    Largest tile = 6 ft

```python
# Step 1: Ask for the length and the width
a = int(input("Length: "))
b = int(input("Width: "))

# Step 2: Use a while loop with Euclid's method (GCD)
while b != 0:
    a, b = b, a % b

# Step 3 & 4: The final value of 'a' is the GCD / largest square tile size
print(f"Largest tile = {a} ft")
```

## 🎫 49. Roll Number Order Check

### Goal:
    A teacher checks whether roll numbers were entered in ascending order.

### Steps:

    Create a list of roll numbers.
    
    Loop through the list and compare each number with the next one.
    
    Use a flag variable that becomes False if any number is out of order.
    
    Print "Sorted" or "Not sorted".

### Example:

    [101, 102, 105, 110] → Sorted
    [101, 105, 103] → Not sorted

```python
# Step 1: Create a list of roll numbers (try changing this to test both cases)
roll_numbers = [101, 102, 105, 110]
# roll_numbers = [101, 105, 103]  # Uncomment this line to test an unsorted list

# Step 2: Initialize a flag variable
is_sorted = True

# Step 3: Loop through the list up to the second-to-last element to compare with the next one
for i in range(len(roll_numbers) - 1):
    # If the current number is greater than the next one, it's out of order
    if roll_numbers[i] > roll_numbers[i + 1]:
        is_sorted = False
        break  # Exit the loop early since we already found it's not sorted

# Step 4: Print the result based on the flag
if is_sorted:
    print(f"{roll_numbers} → Sorted")
else:
    print(f"{roll_numbers} → Not sorted")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
# Step 1: Create the temperatures list
temps = [30, 32, 28, 35, 31, 29, 36]

# Step 2: First loop to add all the temperatures and calculate the average
total_temp = 0
for temp in temps:
    total_temp += temp

average = total_temp / len(temps)

# Step 3: Second loop to count the days above the average
days_above = 0
for temp in temps:
    if temp > average:
        days_above += 1

# Step 4: Print the average (formatted to 2 decimal places) and the count
print(f"Average = {average:.2f}")
print(f"Days above average = {days_above}")
```
