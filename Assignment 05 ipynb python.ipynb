## Create a dictionary that manages records of two patients.

Each patient should have:

    Patient ID
    Name
    Age
    Disease
    Doctor Assigned
    Medicines (a nested dictionary of medicine names and their doses)

### Example:

    Patient 1: ID 501, Name Bilal, Age 45, Disease Diabetes, Doctor Dr. Ayesha
    Medicines: Metformin 500mg, Insulin 10 units
    Patient 2: ID 502, Name Hina, Age 32, Disease Flu, Doctor Dr. Usman
    Medicines: Paracetamol 500mg, Cough Syrup 10ml

### Tasks:

Print the complete patient records.

Print only the medicines of Patient 1.

Add a new medicine "Vitamin D" with dose "1000 IU" for Patient 2.

Change the doctor of Patient 1 to "Dr. Kamran".

Update Patient 2’s age to 33.

Print the final records.

```python
# Create a dictionary managing records of two patients
patient_records = {
    "patient1": {
        "patient_id": 501,
        "name": "Bilal",
        "age": 45,
        "disease": "Diabetes",
        "doctor_assigned": "Dr. Ayesha",
        "medicines": {
            "Metformin": "500mg",
            "Insulin": "10 units"
        }
    },
    "patient2": {
        "patient_id": 502,
        "name": "Hina",
        "age": 32,
        "disease": "Flu",
        "doctor_assigned": "Dr. Usman",
        "medicines": {
            "Paracetamol": "500mg",
            "Cough Syrup": "10ml"
        }
    }
}

# 1. Print the complete patient records
print("--- Complete Patient Records ---")
print(patient_records)
print()

# 2. Print only the medicines of Patient 1
print("--- Medicines of Patient 1 ---")
print(patient_records["patient1"]["medicines"])
print()

# 3. Add a new medicine "Vitamin D" with dose "1000 IU" for Patient 2
patient_records["patient2"]["medicines"]["Vitamin D"] = "1000 IU"

# 4. Change the doctor of Patient 1 to "Dr. Kamran"
patient_records["patient1"]["doctor_assigned"] = "Dr. Kamran"

# 5. Update Patient 2’s age to 33
patient_records["patient2"]["age"] = 33

# 6. Print the final records
print("--- Final Updated Patient Records ---")
print(patient_records)
```

## Create a nested dictionary for a company that stores three employees.

Each employee should have:

    Employee ID
    Name
    Department
    Salary
    Skills (a list of skills inside the dictionary)

Employees:

    ID 201: Name Ali, Department IT, Salary 75000, Skills: Python, SQL, HTML
    ID 202: Name Fatima, Department HR, Salary 60000, Skills: Communication, Recruitment
    ID 203: Name Zain, Department Finance, Salary 80000, Skills: Excel, Accounting

### Tasks:

    Print the complete employee database.
    Print only the skills of employee 201.
    Add a new skill "JavaScript" for Ali.
    Increase Fatima’s salary to 65000.
    Add a new employee: ID 204, Name Hina, Department Marketing, Salary 55000, Skills: SEO, Content Writing.
    Remove employee 203.
    Print the final database.

```python
# Create a nested dictionary for the company storing three employees
company_database = {
    "emp201": {
        "employee_id": 201,
        "name": "Ali",
        "department": "IT",
        "salary": 75000,
        "skills": ["Python", "SQL", "HTML"]
    },
    "emp202": {
        "employee_id": 202,
        "name": "Fatima",
        "department": "HR",
        "salary": 60000,
        "skills": ["Communication", "Recruitment"]
    },
    "emp203": {
        "employee_id": 203,
        "name": "Zain",
        "department": "Finance",
        "salary": 80000,
        "skills": ["Excel", "Accounting"]
    }
}

# 1. Print the complete employee database
print("--- Complete Employee Database ---")
print(company_database)
print()

# 2. Print only the skills of employee 201
print("--- Skills of Employee 201 (Ali) ---")
print(company_database["emp201"]["skills"])
print()

# 3. Add a new skill "JavaScript" for Ali
company_database["emp201"]["skills"].append("JavaScript")

# 4. Increase Fatima’s salary to 65000
company_database["emp202"]["salary"] = 65000

# 5. Add a new employee: ID 204, Name Hina, Department Marketing, Salary 55000, Skills: SEO, Content Writing
company_database["emp204"] = {
    "employee_id": 204,
    "name": "Hina",
    "department": "Marketing",
    "salary": 55000,
    "skills": ["SEO", "Content Writing"]
}

# 6. Remove employee 203
del company_database["emp203"]

# 7. Print the final database
print("--- Final Updated Employee Database ---")
print(company_database)
```

## Common Subjects Between Two Student
Create a set :

    Ali’s subjects: Math, Science, English, Computer, Urdu

    Sara’s subjects: Science, English, Islamiat, Computer, Arts
## Find and print:

    Subjects that both students study (common subjects)
    All subjects taken by either Ali or Sara (combined unique subjects)
    Subjects that only Ali studies
    Subjects that only Sara studies

```python
# Create sets for Ali's and Sara's subjects
ali_subjects = {"Math", "Science", "English", "Computer", "Urdu"}
sara_subjects = {"Science", "English", "Islamiat", "Computer", "Arts"}

# 1. Common subjects (Intersection using .intersection() or &)
common_subjects = ali_subjects.intersection(sara_subjects)
print("Common Subjects:", common_subjects)

# 2. All subjects taken by either Ali or Sara (Union using .union() or |)
all_subjects = ali_subjects.union(sara_subjects)
print("All Unique Subjects:", all_subjects)

# 3. Subjects that only Ali studies (Difference using .difference() or -)
only_ali = ali_subjects.difference(sara_subjects)
print("Subjects only Ali studies:", only_ali)

# 4. Subjects that only Sara studies
only_sara = sara_subjects.difference(ali_subjects)
print("Subjects only Sara studies:", only_sara)
```
