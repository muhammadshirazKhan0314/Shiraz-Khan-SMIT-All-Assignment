# 11. Simple Calculator
# Ask the user for two numbers and an operator (+, -, *, /):
# Perform the corresponding operation.


```python
num1 = float(input("5: "))
num2 = float(input("2r: "))
op = input("Enter operator (+, -, *, /): ")

if op == "+":
    print("Result:", num1 + num2)
elif op == "-":
    print("Result:", num1 - num2)
elif op == "*":
    print("Result:", num1 * num2)
elif op == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero")
else:
    print("Invalid operator")

```

# 26. Traffic System Fine Calculator
# Speed limit = 60 km/h:
# - Speed ≤ 60 → No fine
# - 61–80 → Fine = 500
# - 81–100 → Fine = 2000
# - >100 → License suspended
# But: Emergency vehicle → no fine; Rainy weather AND speed > 60 → fine increases by 50%


```python
speed = float(input("Enter speed (km/h): "))
is_emergency = input("Emergency vehicle? (yes/no): ").lower()
is_rainy = input("Is it rainy weather? (yes/no): ").lower()

if is_emergency == "yes":
    print("No fine (Emergency vehicle)")
else:
    fine = 0
    if speed <= 60:
        print("No fine")
    elif speed <= 80:
        fine = 500
    elif speed <= 100:
        fine = 2000
    else:
        print("License suspended")
        fine = 0

    if speed > 60 and is_rainy == "yes" and speed <= 100:
        fine *= 1.5
    
    if fine > 0:
        print(f"Fine amount: {fine}")
```

# 27. Company Candidate Filter
# - Experience ≥ 5 AND skill score ≥ 80 → shortlisted
# - Experience 2–4 AND skill score ≥ 70 → interview
# - Freshers → only if test score ≥ 85
# - Criminal record = True → reject immediately
# - Multiple switches (>3 in 2 years) → flag as unstable


```python
experience = float(input("Enter years of experience (0 for fresher): "))
criminal_record = input("Criminal record? (True/False): ").capitalize() == "True"

if criminal_record:
    print("Reject immediately (Criminal record)")
else:
    switches = int(input("Number of job switches in last 2 years: "))
    if switches > 3:
        print("Flagged as unstable")
    
    if experience >= 5:
        skill_score = float(input("Enter skill score: "))
        if skill_score >= 80:
            print("Shortlisted")
    elif experience >= 2:
        skill_score = float(input("Enter skill score: "))
        if skill_score >= 70:
            print("Interview")
    else:
        test_score = float(input("Enter test score: "))
        if test_score >= 85:
            print("Shortlisted (Fresher)")
        else:
            print("Not shortlisted")

```

# 28. Passenger Clearance Status
# - baggage > 30kg → extra charge
# - passport invalid → reject
# - visa invalid → reject
# - criminal flagged → security hold
# - otherwise → cleared for boarding


```python
baggage = float(input("Enter baggage weight (kg): "))
passport = input("Passport valid? (Yes/No): ").capitalize()
visa = input("Visa status? (Valid/Invalid): ").capitalize()
criminal = input("Criminal check? (Clear/Flagged): ").capitalize()

if passport != "Valid":
    print("Reject (Invalid Passport)")
elif visa != "Valid":
    print("Reject (Invalid Visa)")
elif criminal == "Flagged":
    print("Security hold (Criminal record flagged)")
else:
    if baggage > 30:
        print("Cleared for boarding (Extra charge for baggage > 30kg)")
    else:
        print("Cleared for boarding")
```

# 30. Suspicious Banking Transactions
# - amount > 100,000 → suspicious
# - location mismatch + high amount → fraud alert
# - night transaction + unverified device → high risk
# - many transactions (>10 in 1 hour) → freeze account
# - otherwise → normal transaction


```python
amount = float(input("Enter transaction amount: "))
location_match = input("Location match? (Yes/No): ").capitalize() == "Yes"
time_of_tx = input("Time of transaction? (Day/Night): ").capitalize()
prev_count = int(input("Previous transactions count in last hour: "))
device_verified = input("Device verified? (Yes/No): ").capitalize() == "Yes"

if prev_count > 10:
    print("Freeze account (Too many transactions)")
elif not location_match and amount > 50000:
    print("Fraud alert (Location mismatch + high amount)")
elif time_of_tx == "Night" and not device_verified:
    print("High risk (Night transaction + unverified device)")
elif amount > 100000:
    print("Suspicious (High amount)")
else:
    print("Normal transaction")
```

# 31. Police Database Risk Level
# - criminal_record = Yes AND cases > 3 → High Risk
# - flagged travel + suspicious financial score → High Risk
# - no activity for long time (>365 days) → Monitor
# - clean record + normal activity → Low Risk
# - mixed signals → Medium Risk


```python
criminal_rec = input("Criminal record? (Yes/No): ").capitalize() == "Yes"
cases = int(input("Number of cases: "))
travel_flagged = input("Travel history flagged? (Yes/No): ").capitalize() == "Yes"
financial_score = float(input("Financial activity score (0-100): "))
last_activity = int(input("Days since last activity: "))

if (criminal_rec and cases > 3) or (travel_flagged and financial_score < 40):
    print("High Risk")
elif last_activity > 365:
    print("Monitor")
elif not criminal_rec and financial_score >= 60 and not travel_flagged:
    print("Low Risk")
else:
    print("Medium Risk")
    
```
